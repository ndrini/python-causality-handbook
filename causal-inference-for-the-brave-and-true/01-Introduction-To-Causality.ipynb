## Rispondere a un tipo diverso di domanda

Il Machine Learning è attualmente molto bravo a rispondere al tipo di domande di tipo predittivo. Come Ajay Agrawal, Joshua Gans e Avi Goldfarb scrivono nel libro Prediction Machines, "la nuova ondata di intelligenza artificiale non ci porta in realtà l'intelligenza, ma una componente critica dell'intelligenza: la predizione". Possiamo fare ogni sorta di cose meravigliose con il machine learning. L'unico requisito è che inquadriamo i nostri problemi come problemi di predizione. Vuoi tradurre dall'inglese al portoghese? Allora costruisci un modello ML che predica frasi in portoghese quando gli vengono date frasi in inglese. Vuoi riconoscere i volti? Allora crea un modello ML che predica la presenza di un volto in una sottosezione di un'immagine. Vuoi costruire un'auto a guida autonoma? Allora crea un modello ML per predire la direzione del volante e la pressione sui freni e sull'acceleratore quando gli vengono presentate immagini e sensori dall'ambiente circostante l'auto.

Tuttavia, il ML non è una panacea. Può fare meraviglie entro confini rigidi e fallire miseramente se i suoi dati deviano un po' da ciò a cui il modello è abituato. Per fare un altro esempio da Prediction Machines, "in molti settori, prezzi bassi sono associati a vendite basse. Per esempio, nel settore alberghiero, i prezzi sono bassi fuori dalla stagione turistica, e i prezzi sono alti quando la domanda è massima e gli hotel sono pieni. Dati questi dati, una predizione ingenua potrebbe suggerire che aumentare il prezzo porterebbe a più camere vendute."

Il ML è notoriamente scarso in questo tipo di problema di causalità inversa. Ci richiedono di rispondere a domande "e se?", che gli economisti chiamano controfattuali. Cosa succederebbe se usassi un altro prezzo invece di questo che sto chiedendo per la mia merce? Cosa succederebbe se facessi una dieta a basso contenuto di zuccheri invece di questa dieta a basso contenuto di grassi? Se lavori in una banca, concedendo credito, dovrai capire come il cambiamento della linea di credito dei clienti influenza i tuoi ricavi. Oppure, se lavori nella pubblica amministrazione, potrebbe esserti chiesto di capire come migliorare il sistema scolastico. Dovresti dare tablet a ogni bambino perché l'era della conoscenza digitale te lo dice? O dovresti costruire una biblioteca alla vecchia maniera?

Al cuore di queste domande c'è un'indagine causale di cui desideriamo conoscere la risposta. Le domande causali permeano i problemi quotidiani, come capire come far aumentare le vendite. Tuttavia, giocano anche un ruolo essenziale in dilemmi che sono molto personali e cari a noi: devo andare in una scuola costosa per avere successo nella vita (l'istruzione causa guadagni)? L'immigrazione riduce le mie possibilità di trovare lavoro (l'immigrazione causa un aumento della disoccupazione)? I trasferimenti di denaro ai poveri riducono il tasso di criminalità? Non importa in quale campo ti trovi. È molto probabile che tu abbia dovuto o dovrai rispondere a qualche tipo di domanda causale. Sfortunatamente per il ML, non possiamo fare affidamento su predizioni di tipo correlazione per affrontarle.

Rispondere a questo tipo di domande è più impegnativo di quanto la maggior parte delle persone apprezzi. I tuoi genitori probabilmente ti hanno ripetuto che "associazione non è causalità", "associazione non è causalità". Ma in realtà, spiegare perché è così è un po' più complesso. Questo è ciò di cui tratta questa introduzione all'inferenza causale. Per quanto riguarda il resto di questo libro, sarà dedicato a **capire come far sì che l'associazione diventi causalità**.

## Quando l'associazione è causalità

Intuitivamente, sappiamo più o meno perché l'associazione non è causalità. Se qualcuno ti dice che le scuole che danno tablet ai loro studenti ottengono risultati migliori di quelle che non lo fanno, puoi rapidamente far notare che probabilmente quelle scuole con i tablet sono più ricche. Di conseguenza, otterrebbero risultati migliori della media anche senza i tablet. Per questo motivo, **non possiamo concludere che dare tablet ai bambini durante le lezioni causerà un aumento delle loro prestazioni** accademiche. **Possiamo solo dire che i tablet a scuola sono associati a prestazioni accademiche elevate**, come misurato dall'ENEM [nota del traduttore: una sorta di esame di maturità in Brasile, che sta per Esame Nazionale della Scuola Superiore]:

```python
import pandas as pd
import numpy as np
from scipy.special import expit
import seaborn as sns
from matplotlib import pyplot as plt
from matplotlib import style

style.use("fivethirtyeight")

np.random.seed(123)
n = 100
tuition = np.random.normal(1000, 300, n).round()
tablet = np.random.binomial(1, expit((tuition - tuition.mean()) / tuition.std())).astype(bool)
# pensa all'enem come all'esame di maturità
enem_score = np.random.normal(200 - 50 * tablet + 0.7 * tuition, 200)
enem_score = (enem_score - enem_score.min()) / enem_score.max()
enem_score *= 1000

data = pd.DataFrame(dict(enem_score=enem_score, Tuition=tuition, Tablet=tablet))
```

```python
plt.figure(figsize=(6,8))
sns.boxplot(y="enem_score", x="Tablet", data=data).set_title('ENEM score by Tablet in Class')
plt.show()
```

Per andare oltre la semplice intuizione, **stabiliamo** prima **una notazione**. Questo sarà il nostro linguaggio quotidiano per parlare di causalità. Pensalo come la lingua comune che useremo per identificare altri coraggiosi e sinceri guerrieri causali, e che comporrà il nostro grido nelle molte battaglie a venire.

Chiamiamo **$T_i$ l'assunzione del trattamento per l'unità i**. [nota del traduttore: nell'esempio l'**unità i** rappresenta la **singola scuola**.]


$
T_i=\begin{cases}
1 \ \text{se l'unità i ha ricevuto il trattamento}\\
0 \ \text{altrimenti}\\
\end{cases}
$

Il **trattamento** qui non deve essere necessariamente un medicinale o qualcosa dal campo medico. È semplicemente un termine che useremo per indicare **un qualche intervento** di cui vogliamo conoscere l'effetto. Nel nostro caso, il trattamento è dare tablet agli studenti. Come nota a margine, potresti talvolta vedere $D$ invece di $T$ per indicare il trattamento.

Ora, chiamiamo **$Y_i$ la variabile di risultato osservata** per l'unità i.

Il risultato è la nostra variabile di interesse. Vogliamo sapere se il trattamento ha qualche influenza su di essa. Nel nostro esempio dei tablet, sarebbe la prestazione accademica.

Ecco dove le cose si fanno interessanti. Il **problema fondamentale dell'inferenza causale** è che non possiamo mai osservare la stessa unità con e senza trattamento. È come se avessimo due strade divergenti e potessimo sapere cosa c'è davanti solo su quella che prendiamo. Come nella poesia di Robert Frost:

>Due strade divergevano in un bosco giallo,  
E dispiaciuto di non poterle percorrere entrambe  
Ed essere un solo viaggiatore, a lungo rimasi  
E guardai lungo una fin dove potevo  
Là dove si perdeva nel sottobosco;  


Per comprendere questo concetto, parleremo molto in termini di **risultati potenziali** (potential outcomes). Sono potenziali perché **non sono realmente accaduti**. Invece indicano **cosa sarebbe successo** nel caso in cui un certo trattamento fosse stato somministrato. A volte chiamiamo il risultato potenziale che è accaduto, fattuale, e quello che non è accaduto, controfattuale.

Per quanto riguarda la notazione, usiamo un pedice aggiuntivo:

$Y_{0i}$ è il risultato potenziale per l'unità i senza il trattamento.

$Y_{1i}$ è il risultato potenziale per **la stessa** unità i con il trattamento.

A volte potresti vedere i risultati potenziali rappresentati come funzioni $Y_i(t)$, quindi fai attenzione. $Y_{0i}$ potrebbe essere $Y_i(0)$ e $Y_{1i}$ potrebbe essere $Y_i(1)$. Qui useremo la notazione con pedice nella maggior parte dei casi.

![img](./data/img/intro/potential_outcomes.png)_

[n.d.t. se vi è il trattamento, allora sto prendendo la strata $Y_1$]

Tornando al nostro esempio, $Y_{1i}$ è la prestazione accademica per lo studente i se si trova in una classe con i tablet. Che sia o meno il caso, non importa per $Y_{1i}$. È lo stesso indipendentemente. Se lo studente i riceve il tablet, possiamo osservare $Y_{1i}$. Se no, possiamo osservare $Y_{0i}$. Nota come in quest'ultimo caso, $Y_{1i}$ è ancora definito, semplicemente non possiamo vederlo. In questo caso, è un risultato potenziale controfattuale.

**Con i risultati potenziali, possiamo definire l'effetto individuale del trattamento**:

 $Y_{1i} - Y_{0i}$

Naturalmente, a causa del problema fondamentale dell'inferenza causale, non possiamo mai conoscere l'effetto individuale del trattamento perché osserviamo solo uno dei risultati potenziali [nota del traduttore: probabilmente si vuole dire osserviamo un risultato mentre l'altro è mancante e deve essere sostituito con un risultato potenziale]. Per il momento, concentriamoci su qualcosa di più facile rispetto alla stima dell'effetto individuale del trattamento. Invece, concentriamoci sull'**effetto medio del trattamento** (Average Treatment Effect), definito come segue.

$ATE = E[Y_1 - Y_0]$

dove `E[...]` è il valore atteso.

[nota del traduttore: il valore atteso è la media della popolazione, con proprietà di linearità, cioè E[aX + b Y] = aE[X] + bE[Y]. ]

Un'altra quantità più facile da stimare è l'**effetto medio del trattamento sui trattati** (Average Treatment Effect on the Treated):

$ATT = E[Y_1 - Y_0 | T=1]$

[nota del traduttore: in questo caso il valore atteso è soggetto alla condizione che il trattamento sia stato somministrato. Nell'esempio, si usano i dati solo delle scuole a cui i tablet sono stati forniti.]

Ora, so che **non possiamo vedere entrambi i risultati** potenziali, ma solo per amor di discussione, **supponiamo di poterlo fare**. Fingiamo che la divinità dell'inferenza causale sia compiaciuta delle molte battaglie statistiche che abbiamo combattuto e ci abbia ricompensato con poteri divini per vedere i risultati potenziali alternativi. Con quel potere, diciamo che **raccogliamo dati su 4 scuole**. Sappiamo se hanno dato tablet ai loro studenti e il loro punteggio in alcuni test accademici annuali. Qui, **i tablet sono il trattamento, quindi $T=1$ se la scuola fornisce tablet ai suoi studenti. $Y$ sarà il punteggio del test**.

```python
pd.DataFrame(dict(
    i= [1,2,3,4],
    Y0=[500,600,800,700],
    Y1=[450,600,600,750],
    T= [0,0,1,1],
    Y= [500,600,600,750],
    TE=[-50,0,-200,50],
))
```

L'**$ATE$** qui sarebbe la **media dell'ultima colonna**, cioè dell'effetto del trattamento:

$ATE=(-50 + 0 - 200 + 50)/4 = -50$

Questo significherebbe che **i tablet hanno ridotto le prestazioni** accademiche degli studenti, in media, di 50 punti. L'$ATT$ qui sarebbe la media dell'ultima colonna quando $T=1$:

$ATT=(- 200 + 50)/2 = -75$

Questo sta dicendo che, **per le scuole che sono state trattate, i tablet hanno ridotto le prestazioni** accademiche degli studenti, in media, di 75 punti. 


Naturalmente non possiamo mai saperlo. **In realtà, la tabella sopra apparirebbe così**:

```python
pd.DataFrame(dict(
    i= [1,2,3,4],
    Y0=[500,600,np.nan,np.nan],
    Y1=[np.nan,np.nan,600,750],
    T= [0,0,1,1],
    Y= [500,600,600,750],
    TE=[np.nan,np.nan,np.nan,np.nan],
))
```

[nota del traduttore: per le scuole in cui T=0 ho solo $Y_0$, mentre per le scuole in cui T=1, ho solo lo $Y_1$]

---

Questo [n.d.t. l'assenza di dati] non è certo l'ideale, potresti dire, ma **non posso comunque prendere la media dei trattati e confrontarla con la media dei non trattati?** In altre parole, non posso semplicemente fare $ATE=(600+750)/2  - (500 + 600)/2  = 125$? Beh, **no!** Nota quanto sono diversi i risultati. Hai appena commesso il peccato più grave di **confondere l'associazione con la causalità**. Per capire perché, esaminiamo il nemico principale dell'inferenza causale.

[n.d.t. l'associazione dice: T = 1 (alto) è probabile che anche Y sia alto, mentre la causalità dice: T = 1 (alto) causa che Y sia ... in questo caso basso!]   


## Distorsione (Bias)

La distorsione è ciò che rende l'associazione diversa dalla causalità. Fortunatamente, può essere facilmente compresa con la nostra intuizione. Ricapitoliamo il nostro esempio dei tablet in classe. Quando ci confrontiamo con l'**affermazione che le scuole che danno tablet ai loro studenti ottengono punteggi più alti** nei test, possiamo **confutarla dicendo che quelle scuole probabilmente otterrebbero punteggi più alti comunque**, anche senza i tablet. Questo perché probabilmente hanno più soldi delle altre scuole; quindi possono pagare insegnanti migliori, permettersi aule migliori, ecc. In altre parole, le **scuole trattate (con tablet) non sono comparabili con le scuole non trattate**.

Usando la notazione dei risultati potenziali, diremmo che **$Y_0$ dei trattati è diverso da $Y_0$ dei non trattati**. Ricorda che $Y_0$ dei trattati **è controfattuale**. Non possiamo osservarlo, ma possiamo ragionarci sopra. In questo caso particolare, possiamo persino sfruttare la nostra comprensione di come funziona il mondo per andare oltre. Possiamo dire che, probabilmente, $Y_0$ dei trattati è più grande di $Y_0$ delle scuole non trattate. Questo perché le scuole che possono permettersi di dare tablet ai loro studenti possono anche permettersi altri fattori che contribuiscono a punteggi migliori nei test. Lascia che questo concetto si sedimenti per un momento. Ci vuole un po' di tempo per abituarsi a parlare di risultati potenziali. Rileggi questo paragrafo e assicurati di averlo capito.

Con questo in mente, possiamo **mostrare con matematica elementare perché l'associazione non è causalità**. L'associazione è misurata da $E[Y|T=1] - E[Y|T=0]$. 

[n.d.t. il valore più probabile di quando si ha il trattamento meno il valore più probabile quando non si ha il trattamento, che è lo stesso che dire il valore più probabile dei risultati scolastici per le scuole con tablet meno il valore più probabile delle scuole senza tablet.]

Nel nostro esempio, questa è la media dei punteggi per le scuole con tablet meno la media dei punteggi per quelle senza. D'altra parte, la causalità è misurata da $E[Y_1 - Y_0]$.

[n.d.t. il valore più probabile di quando si ha il trattamento meno il valore più probabile di quando non si ha il trattamento: ma uno dei due valori non è noto sperimentalmente ma è controfattuale!]

Prendiamo la misura dell'associazione e sostituiamo i risultati osservati con i risultati potenziali per vedere come si relazionano. Per i trattati, il risultato osservato è $Y_1$. Per i non trattati, il risultato osservato è $Y_0$.

$
E[Y|T=1] - E[Y|T=0] = E[Y_1|T=1] - E[Y_0|T=0]
$

[n.d.t. nell'espressione E[Y∣T=1], il condizionamento ∣T=1 seleziona automaticamente lo scenario 1. Ecco perché Y "diventa" $Y_1$​.]

Ora, aggiungiamo e sottraiamo $E[Y_0|T=1]$. Questo è un risultato controfattuale. Ci dice quale sarebbe stato il risultato dei trattati se non avessero ricevuto il trattamento.

$
E[Y|T=1] - E[Y|T=0] = E[Y_1|T=1] - E[Y_0|T=0] + E[Y_0|T=1] - E[Y_0|T=1]
$

Infine, riordiniamo i termini, uniamo alcune aspettative, ed ecco:

[n.d.t. uniamo E[Y_1|T=1] -  E[Y_0|T=1] , data la linearità di E in E[Y_1 - Y_0|T=1]]

$
E[Y|T=1] - E[Y|T=0] = \underbrace{E[Y_1 - Y_0|T=1]}_{ATT} + \underbrace{\{ E[Y_0|T=1] - E[Y_0|T=0] \}}_{BIAS}
$

[n.d.t. il secondo termine, il bias, dice qual è la differenza fra $E[Y_0|T=1]$ (quanto sarebbeil risultato di quelle suole che hanno avuto il trattamento se non lo avessero avuto) meno 
 $E[Y_0|T=0] (quanto è il risultato per chi non ha avuto il trattamento, nel nostro esempio per le scuole che non hanno avuto il tablet).
Tradotto **in linguaggio comune**, il bias risponde alla domanda:
"Se nessuno ricevesse i tablet, **le scuole dei due gruppi sarebbero comunque diverse**?" ]


Questo semplice pezzo di matematica racchiude tutti i problemi che incontreremo nelle domande causali. Non posso sottolineare abbastanza quanto sia importante comprenderne ogni aspetto. Se mai fossi costretto a tatuarti qualcosa sul braccio, questa equazione sarebbe una buona candidata. È qualcosa da tenere caro e comprendere cosa ci sta dicendo, come un testo sacro che può essere interpretato in 100 modi diversi. Infatti, diamo un'occhiata più approfondita. Scomponiamola in alcune delle sue implicazioni. Primo, questa equazione ci dice perché l'associazione non è causalità. Come possiamo vedere, l'associazione è uguale all'effetto del trattamento sui trattati più un termine di distorsione. **La distorsione è data da quanto il gruppo trattato e il gruppo di controllo differiscono prima del trattamento, nel caso in cui nessuno dei due abbia ricevuto il trattamento**. Ora possiamo dire precisamente perché siamo sospettosi quando qualcuno ci dice che i tablet in classe migliorano le prestazioni accademiche. Pensiamo che, in questo esempio, $E[Y_0|T=0] < E[Y_0|T=1]$, cioè, le scuole che possono permettersi di dare tablet ai loro studenti sono migliori di quelle che non possono, **indipendentemente dal trattamento dei tablet**.




Perché succede questo? Ne parleremo di più quando affronteremo il confondimento, ma per ora, puoi pensare alla distorsione come qualcosa che nasce perché molte cose che non possiamo controllare cambiano insieme al trattamento. Di conseguenza, le scuole trattate e non trattate non differiscono solo per i tablet. Differiscono anche per il costo della retta, la posizione, gli insegnanti...
Per poter dire che i tablet in classe aumentano le prestazioni accademiche, avremmo bisogno che le scuole con e senza di essi fossero, in media, simili tra loro.

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x="Tuition", y="enem_score", hue="Tablet", data=data, s=70).set_title('ENEM score by Tuition Cost')
plt.show()
```

Now that we understand the problem let's look at the solution. We can also say what would be necessary to make association equal to causation. **If $E[Y_0|T=0] = E[Y_0|T=1]$, then, association IS CAUSATION!** Understanding this is not just remembering the equation. There is a strong intuitive argument here. To say that $E[Y_0|T=0] = E[Y_0|T=1]$ is to say that treatment and control group are comparable before the treatment. Or, when the treated had not been treated, if we could observe its $Y_0$, its outcome would be the same as the untreated. Mathematically, the bias term would vanish:

$
E[Y|T=1] - E[Y|T=0] = E[Y_1 - Y_0|T=1] = ATT
$

Also, if the treated and the untreated only differ on the treatment itself, then, $E[Y_0|T=0] = E[Y_0|T=1]$ and we have that the causal impact on the treated is the same as in the untreated (because they are very similar).

$
\begin{align}
E[Y_1 - Y_0|T=1] &= E[Y_1|T=1] - E[Y_0|T=1] \\
&= E[Y_1|T=1] - E[Y_0|T=0] \\
&= E[Y|T=1] - E[Y|T=0]
\end{align}
$

In this case, the **difference in means BECOMES the causal effect**:

$
E[Y|T=1] - E[Y|T=0] = ATT
$

Additionally, if the treated and the untreated only differ on the treatment itself, we also have $E[Y_1|T=0] = E[Y_1|T=1]$, that is, we make sure that both treated and control groups respond similarly to the treatment. Now, besides being exchangeable prior to the treatment, treated and untreated are also exchangeable **after the treatment**. In this case, $E[Y_1 - Y_0|T=1]=E[Y_1 - Y_0|T=0]$ and
 
$
E[Y|T=1] - E[Y|T=0] = ATT = ATE
$

Once again, this is so important that I think it is worth going over it again, now with pretty pictures. If we make a simple average comparison between the treatment and the untreated group, this is what we get (blue dots didn't receive the treatment, that is, the tablet):

![img](./data/img/intro/anatomy1.png)

Notice how the difference in outcomes between the two groups can have two causes:

1. The treatment effect. The increase in test scores is caused by giving kids tablets.
2. Some of the differences in test scores can be due to tuition prices for better education. In this case, treated and untreated differ because the treated have a much higher tuition price. Other differences between the treatment and untreated are NOT the treatment itself. 

The individual treatment effect is the difference between the unit's outcome and another theoretical outcome that the same unit would have if it got the alternative treatment. The actual treatment effect can only be obtained if we have godlike powers to observe the potential outcome, like the left figure below. These are the counterfactual outcomes and are denoted in a light color.

![img](./data/img/intro/anatomy2.png)

In the right plot, we depicted the bias that we’ve talked about before. We get the bias if we set everyone to not receive the treatment. In this case, we are only left with the $T_0$ potential outcome. Then, we see how the treated and untreated groups differ. If they do, something other than the treatment is causing the treated and untreated to be different. This something is the bias and is what shadows the actual treatment effect.

Now, contrast this with a hypothetical situation where there is no bias. Suppose that tablets are randomly assigned to schools. In this situation, rich and poor schools have the same chance of receiving the treatment. Treatment would be well distributed across the tuition spectrum.

![img](./data/img/intro/anatomy3.png)

In this case, the difference in the outcome between treated and untreated IS the average causal effect. This happens because there is no other source of difference between treatment and untreated other than the treatment itself. All the differences we see must be attributed to it. Another way to say this is that there is no bias.

![img](./data/img/intro/anatomy4.png)

If we set everyone to not receive the treatment so that we only observe the $Y_0$s, we would find no difference between the treated and untreated groups. 

This is the herculean task causal inference is all about. It is about finding clever ways of removing bias and making the treated and the untreated comparable so that all the difference we see is only the average treatment effect. Ultimately, causal inference is about figuring out how the world works, stripped of all delusions and misinterpretations. And now that we understand this, we can move forward to mastering some of the most powerful methods to remove bias, the weapons of the Brave and True, to identify the causal effect. 

## Key Ideas

But unfortunately, we can only measure one of them, where the fundamental problem of causal inference lies.
